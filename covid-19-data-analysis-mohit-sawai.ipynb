```python
covid_df.drop(["Sno", "Time", "ConfirmedIndianNational","ConfirmedForeignNational"], inplace = True, axis =1)
```

```python
covid_df.head(5)
```

```python
covid_df['Date'] = pd.to_datetime(covid_df['Date'], format='%Y-%m-%d')
```

```python
covid_df.head()
```

```python
#Active Cases

covid_df['Active_Cases'] = covid_df['Confirmed'] - (covid_df['Cured'] + covid_df['Deaths'])
```

```python
covid_df.head(588)
```

```python
#Pivot_Table
statewise = pd.pivot_table(covid_df, values = ["Confirmed", "Deaths", "Cured"],
                           index = "State/UnionTerritory")
```

```python
statewise["Recovery Rate"] = statewise["Cured"]*100/statewise["Confirmed"]
```

```python
statewise["Mortality Rate"] = statewise["Deaths"]*100/statewise["Confirmed"]
```

```python
statewise= statewise.sort_values(by = "Confirmed", ascending = False)
```

```python
statewise.style.background_gradient(cmap = "cubehelix")
```

```python
#Top 10 Active Cases States
top_10_active_cases = covid_df.groupby(by = 'State/UnionTerritory').max()[['Active_Cases', 'Date' ]].sort_values(by = ['Active_Cases'], ascending = False).reset_index()
```

```python
fig = plt.figure(figsize= (16,9))
```

```python
plt.title("Top 10 states with most active cases in India", size = 25)
```

```python
ax = sns.barplot(data = top_10_active_cases.iloc[:10], y = "Active_Cases", x = "State/UnionTerritory", linewidth = 2, edgecolor = 'red')
```

```python
#Top 10 Active Cases States
top_10_active_cases = covid_df.groupby(by = 'State/UnionTerritory').max()[['Active_Cases', 'Date' ]].sort_values(by = ['Active_Cases'], ascending = False).reset_index()
fig = plt.figure(figsize= (16,9))
plt.title("Top 10 states with most active cases in India", size = 25)
ax = sns.barplot(data = top_10_active_cases.iloc[:10], y = "Active_Cases", x = "State/UnionTerritory", linewidth = 2, edgecolor = 'black')
plt.xlabel('states')
plt.ylabel("Total Active Cases")
plt.show()
```

```python
#Top states with highest deaths
top_10_death_cases = covid_df.groupby(by = 'State/UnionTerritory').max()[['Deaths', 'Date' ]].sort_values(by = ['Deaths'], ascending = False).reset_index()
fig = plt.figure(figsize= (18,5))
plt.title("Top 10 states with most Deaths", size = 25)
ax = sns.barplot(data = top_10_death_cases.iloc[:12], y = "Deaths", x = "State/UnionTerritory", linewidth = 2, edgecolor = 'red')
plt.xlabel('states')
plt.ylabel("Total Deaths")
plt.show()
```

```python
# growth trend
fig = plt.figure(figsize=(10,6))

states = ['Maharashtra', 'Karnataka', 'Kerala', 'Tamil Nadu', 'Uttar Pradesh']

ax = sns.lineplot(
    data=covid_df[covid_df['State/UnionTerritory'].isin(states)],
    x='Date',
    y='Active_Cases',
    hue='State/UnionTerritory'
)

ax.set_title("Top 5 Affected States in India", size=16)
plt.show()


```

```python
# Remove rows where State = India
vaccine = vaccine_df[vaccine_df.State != 'India']
vaccine
```

```python
vaccine.rename(columns={"Total Individuals Vaccinated": "Total"}, inplace=True)
vaccine.head() 
```

```python
#Most vaccined State

max_vac = vaccine.groupby('State')['Total'].sum().to_frame('Total')
max_vac = max_vac.sort_values('Total', ascending = False) [:5]
max_vac
```

```python
# Top states with highest vaccination
most_vaccined_state = vaccine.groupby(by='State').max()[['Total']].sort_values(by=['Total'], ascending=False).reset_index()

fig = plt.figure(figsize=(15, 5))
plt.title("Top 5 Most Vaccinated States", size=20)
ax = sns.barplot(data=most_vaccined_state.iloc[:10], y="Total", x="State", linewidth=2, edgecolor='red')
plt.xlabel('States')
plt.ylabel("Vaccination")
plt.show()
```

```python
# Top states with least vaccination
least_vaccined_state = vaccine.groupby(by='State').max()[['Total']].sort_values(by=['Total'], ascending=True).reset_index()

fig = plt.figure(figsize=(30, 5))
plt.title("Top 10 Least Vaccinated States", size=20)
ax = sns.barplot(data=least_vaccined_state.iloc[:10], y="Total", x="State", linewidth=2, edgecolor='red')
plt.xlabel('States')
plt.ylabel("Vaccination")
plt.show()
```
